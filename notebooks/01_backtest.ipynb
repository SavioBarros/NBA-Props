# Backtest de pontos da NBA
Compare a média recente com a projeção ponderada usando apenas jogos anteriores à data prevista. A avaliação de probabilidades usa uma linha simulada definida antes de cada jogo. A demonstração usa um histórico fictício.


```python
from pathlib import Path
import sys

root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
sys.path.insert(0, str(root))

from nba_history import load_logs
from nba_props_backtest import backtest_points, demo_history, plot_report, summarize

demo = True
window = 15
weight = 0.65
games = demo_history() if demo else load_logs(root / 'data/nba_history.sqlite')
predictions = backtest_points(games, window=window, weight=weight)
predictions.head()
```

```python
metrics, calibration = summarize(predictions)
metrics.round(3)
```

```python
plot_report(metrics, calibration, demo=demo).show()
```
